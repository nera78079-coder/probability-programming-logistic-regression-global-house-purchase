```python
import pandas as pd
import pymc as pm
import arviz as az
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder
```

```python
df=pd.read_csv("global_house_purchase_dataset.csv")
df.head()
```

```python
cat_nom=['country', 'city', 'property_type']
cat_nom
```

```python
cat_ord=["furnishing_status"]
cat_ord
```

```python
for i in cat_nom:
    enco=LabelEncoder()
    df.drop(columns=i)
    df[i]=enco.fit_transform(df[i])
```

```python
df[['country', 'city', 'property_type']].head()
```

```python
for i in cat_ord:
    enco=OrdinalEncoder()
    df.drop(columns=i)
    df[i]=enco.fit_transform(df[[i]])
```

```python
df[["furnishing_status"]].head()
```

```python
x=df[['country', 'city', 'property_type', 'furnishing_status',
       'property_size_sqft', 'price', 'constructed_year', 'previous_owners',
       'rooms', 'bathrooms', 'garage', 'garden', 'crime_cases_reported',
       'legal_cases_on_property', 'customer_salary', 'loan_amount',
       'loan_tenure_years', 'monthly_expenses', 'down_payment',
       'emi_to_income_ratio', 'satisfaction_score', 'neighbourhood_rating',
       'connectivity_score']]
y=df['decision']
```

```python
with pm.Model() as model:
    
    cols = list(x.columns)              
    X = x.values                        

    w = []
    for col in cols:
        w.append(pm.Normal(col, mu=0, sigma=1))

    y_theo=w[0] * X[:, 0]

    for idx in range(1, len(w)):
        y_theo =  y_theo+ w[idx] * X[:, idx]

    y_pred = pm.Bernoulli("y", logit_p=y_theo, observed=y)

    trace = pm.sample(2000, tune=1000, target_accept=0.9)


```
